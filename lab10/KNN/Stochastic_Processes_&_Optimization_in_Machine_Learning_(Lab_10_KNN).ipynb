**Briefly describe how KNN algorithm works.**

KNN is a supervised, non-parametric classification algorithm. During the training phase, KNN does not build an explicit model or estimate parameters. Instead, it simply stores the $N$ labeled training vectors together with their class labels.

During the prediction phase, a new unlabeled vector $\hat{{x}}$
is compared with all stored training vectors using a distance metric, such as Euclidean distance. The algorithm selects the $K$ closest training samples and assigns $\hat{{x}}$ to the class that appears most frequently among these K nearest neighbors.


**What are the main advantages and disadvantages of this algorithm?**

Advantages:

1. Simple and intuitive: KNN is easy to understand and implement. It classifies a new point based on the labels of the closest training examples.
2. No explicit training model: KNN is a lazy learning method. During training, it mainly stores the labeled training vectors instead of estimating model parameters.
3. Non-parametric method: KNN does not assume a specific probabilistic model or functional form for the data. Therefore, it can handle complex and nonlinear classification boundaries.
4. Works well for small or well-structured datasets: If the features are meaningful and properly normalized, KNN can give good results.

Disadvantages:

1. High prediction cost: For every new test point, KNN must compute the distance to all training points. Therefore, prediction can be expensive.
2. High storage requirement: Since KNN stores the whole training dataset, it requires more memory when the dataset is large.
3. Sensitive to feature scaling: Features with larger numerical ranges can dominate the distance calculation. For this reason, normalization is important.
4. Sensitive to the choice of K: A very small $K$, such as $K=1$, can make the classifier sensitive to noise. A very large $K$ can smooth the decision boundary too much and ignore local structure.
5. Less efficient in high-dimensional data: When the number of features is large, distances become less meaningful.

**Is KNN efficient when the dataset has a large number of features? What happens when we choose large or small values for k?**

KNN is usually not very efficient when the dataset has many features. During prediction, the algorithm must compute the distance between the new test point and all stored training points. If there are N training samples and m features, the prediction cost is approximately O(Nm).

Also, in high-dimensional spaces, distances become less meaningful. Many points may appear similarly far from each other, so the idea of a “nearest neighbor” becomes weaker. This is known as the curse of dimensionality. For this reason, KNN often requires feature normalization, dimensionality reduction, or removal of irrelevant features/outliers.

If $K$ is very small, for example K=1, the classifier assigns the new point to the class of only its nearest neighbor. This can make the model very sensitive to noise and outliers, so it may overfit.

If $K$ is large, the classifier uses more neighbors and becomes more stable and tolerant to noise. However, if $K$ is too large, it may smooth the decision boundary too much, ignore local structure, and lead to underfitting.

```python
# Make Predictions with k-nearest neighbors on the Iris Flowers Dataset
from csv import reader
from math import sqrt
```

```python
# Load a CSV file
def load_csv(filename):
	dataset = list()
	with open(filename, 'r') as file:
		csv_reader = reader(file)
		for row in csv_reader:
			if not row:
				continue
			dataset.append(row)
	return dataset

# Convert string column to float
def str_column_to_float(dataset, column):
	for row in dataset:
		row[column] = float(row[column].strip())

# Convert string column to integer
def str_column_to_int(dataset, column):
	class_values = [row[column] for row in dataset]
	unique = set(class_values)
	lookup = dict()
	for i, value in enumerate(unique):
		lookup[value] = i
		print('[%s] => %d' % (value, i))
	for row in dataset:
		row[column] = lookup[row[column]]
	return lookup

# Find the min and max values for each column
def dataset_minmax(dataset):
	minmax = list()
	for i in range(len(dataset[0])):
		col_values = [row[i] for row in dataset]
		value_min = min(col_values)
		value_max = max(col_values)
		minmax.append([value_min, value_max])
	return minmax

# Rescale dataset columns to the range 0-1
def normalize_dataset(dataset, minmax):
	for row in dataset:
		for i in range(len(row)):
			row[i] = (row[i] - minmax[i][0]) / (minmax[i][1] - minmax[i][0])
```

* Mention other methods for calculating distance.

Manhattan Distance:  $d_{Manhattan}​(a,b)=\sum\limits_{k=1}^{m}​∣a_k​−b_k​|$

Sum of absolute differences. Used for continuous data; cheaper than Euclidean (no squaring/root); more robust to outliers; often better in high dimensions.

Hamming Distance: Number of opposite binary digits between $(\mathbf{x}(i), \mathbf{x}(j))$

Designed for discrete/categorical/binary data — DNA, text, bit strings. Not suitable for continuous features like Iris measurements.


```python
# Calculate the Euclidean distance between two vectors
def euclidean_distance(row1, row2):
	distance = 0.0
	for i in range(len(row1)-1):
		distance += (row1[i] - row2[i])**2
	return sqrt(distance)

# Locate the most similar neighbors
def get_neighbors(train, test_row, num_neighbors):
	distances = list()
	for train_row in train:
		dist = euclidean_distance(test_row, train_row)
		distances.append((train_row, dist))
	distances.sort(key=lambda tup: tup[1])
	neighbors = list()
	for i in range(num_neighbors):
		neighbors.append(distances[i][0])
	return neighbors

# Make a prediction with neighbors
def predict_classification(train, test_row, num_neighbors):
	neighbors = get_neighbors(train, test_row, num_neighbors)
	output_values = [row[-1] for row in neighbors]
	prediction = max(set(output_values), key=output_values.count)
	return prediction
```
